# V1-P5: Final-Test Governance Gate

The single-use, pre-registered confirmatory evaluation against the locked final-test partition (`DEC-062`). F1 is refit once on the full development partition (train plus validation) reporting raw probabilities; operating-point thresholds for the `DEC-061` 2.5% and 5% rates are frozen from that model's own development-partition predictions before the final-test partition is read. This is a governance gate, not an experiment, and it carries no `EXP-` identifier. The result is a confirmatory sanity check, not a performance claim, and this partition may not be read again without a superseding decision.
